```python
import pandas as pd
```

```python
df = pd.read_csv("Data/cancer_detection_ml_practice_1000.csv")
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df["Cancer"] = encoder.fit_transform(df["Cancer_Detected"])
df["Is_Alcohol"] = encoder.fit_transform(df["Alcohol"])
df["Is_Existing_Disease"] = encoder.fit_transform(df["Any_Existing_Disease"])
```
